## 1. Business Understanding
The objectives and requirements of the project. 

### Business objectives
Determine whether cell-nucleus measurements from an FNA image can reliably distinguish malignant from benign breast masses, in order to support faster, more consistent diagnostic triage

### Assess about situation
##### Inventory of resources
- Data: 569 records, 30 numeric features + ID + Diagnosis, no missing values, sourced from UCI (or scikit-learn's bundled copy)
- Tools: Python, pandas, scikit-learn, matplotlib/seaborn, Jupyter
- Domain knowledge: general understanding of what the measurements represent (nucleus radius, texture, etc.), but no clinical/oncology expertise — that's a limitation, not a resource

##### Requirements, assumptions, and constraints
- Requirement: binary classifier (M/B) using kNN specifically, evaluated with hold-out validation
- Assumption: the 569 samples are representative of the broader population of breast masses seen in practice (this is actually a fairly strong assumption — the data comes from a single research group's imaging process, so generalization to other imaging setups/hospitals isn't guaranteed)
- Constraint: this is an academic/demonstration exercise, not a validated clinical tool — nothing built here could be deployed in a real diagnostic setting without much more rigorous clinical validation, regulatory approval, and testing on external data

##### Risks and contingencies
- Class imbalance: check the M/B split; if skewed, accuracy alone can be misleading (a model could get high accuracy by mostly predicting the majority class) — mitigate by also reporting precision/recall per class
- Overfitting to this specific dataset: 569 samples is fairly small; hold-out or cross-validation results may not generalize — mitigate with careful validation and modest k search rather than exhaustive tuning
- Feature scale differences: features like area are numerically much larger than smoothness — since kNN is distance-based, this would bias results toward large-scale features if left unaddressed — mitigate with standardization (already required in your task)
- Multicollinearity: radius, perimeter, and area are geometrically related and likely highly correlated — this doesn't break kNN the way it can break linear models, but it's worth noting in EDA

##### Costs and benefits
- Cost: relatively low — public dataset, standard open-source tools, modest compute (kNN is cheap to train)
- Benefit: a working proof-of-concept for how measurement-based classification could support diagnostic triage, plus (for you) a concrete demonstration of the full CRISP-DM workflow and kNN mechanics

##### Terminology
- Malignant (M) = cancerous; Benign (B) = non-cancerous
- FNA = fine needle aspirate, the biopsy technique used to obtain the cell sample
- kNN = k-Nearest Neighbors, a non-parametric, distance-based classifier

### Data mining goals
Train and evaluate a kNN classifier on the WDBC dataset to predict Diagnosis (M/B) from the 30 numeric features, and determine what level of accuracy/precision/recall it achieves.

### Project plan
1. Business understanding
    
    Define objective (predict malignant vs benign from FNA measurements) and success criteria (prioritize recall on malignant class).
2. Data Understanding
    
    Load WDBC data, check shape/types/missing values, describe variables, check class balance, visualize key features.
3. Data Preparation

    Drop ID, encode Diagnosis (M=1/B=0), split X/y, hold-out train/test split (stratified), standardize features (fit on train only).
4. Modeling

    Train kNN classifier; try several k values (e.g. 1,3,5,7,11,15) via cross-validation on the training set; select best k.
5. Evaluation

    Predict on test set; report confusion matrix, accuracy, precision, recall; compare k choices; interpret errors.
6. Deployment

    Summarize findings, discuss practical use as decision-support, note limitations, suggest next steps.
